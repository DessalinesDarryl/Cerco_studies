```python
import sys
import os

# Ajoute le dossier src/ au path si besoin
sys.path.append(os.path.abspath("../src"))

from cohort import load_patient_groups, group_edf_files_by_pathology
import pandas as pd

# Chargement du fichier Excel
df, group_map, demographics_map = load_patient_groups("../data/Tableau_synthese_patients0.xlsx",  sheet_index=0)

# Fusionne les deux dictionnaires
full_info_map = {}

for patient_id in group_map:
    full_info_map[patient_id] = {
        'label': group_map[patient_id]
    }
    # Ajoute les infos démographiques si disponibles
    if patient_id in demographics_map:
        full_info_map[patient_id].update(demographics_map[patient_id])

print(full_info_map)

# Sauvegarde du dictionnaire complet dans un fichier CSV
df_full_info = pd.DataFrame.from_dict(full_info_map, orient='index')
df_full_info.index.name = 'id_patient'
df_full_info.reset_index(inplace=True)
df_full_info.to_csv("../data/patient_info.csv", index=False)   
```
